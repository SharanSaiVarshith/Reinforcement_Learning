```python
import random

# Part A – Machine Initialization
machines = ['A', 'B', 'C']
N = {m: 0 for m in machines}   # Number of plays per machine
R = {m: [] for m in machines}  # Rewards history
Q = {m: 0.0 for m in machines} # Estimated Q-values

# Parameters
N_play = 10
epsilon = 0.2

print("Modified ε-Greedy Strategy for Multi-Armed Bandit\n")

for play in range(1, N_play + 1):
    # Part C – Decision
    r = random.random()
    best_machine = max(Q, key=Q.get)

    if r < epsilon:  # Exploration
        choices = [m for m in machines if m != best_machine]
        selected = random.choice(choices)
        print(f"Play {play}: Exploration → Selected {selected}")
    else:            # Exploitation
        selected = best_machine
        print(f"Play {play}: Exploitation → Selected {selected}")

    # Part D – Reward Collection
    reward = int(input(f"Enter reward for machine {selected}: "))
    R[selected].append(reward)

    # Part E – Update counts and Q-value
    N[selected] += 1
    Q[selected] += (reward - Q[selected]) / N[selected]

    print(f"Updated Q({selected}) = {Q[selected]:.2f}\n")

# Part F – Display Results
print("\nFinal Results after 10 plays:")
print("Machine | Count | Rewards | Q-value")
for m in machines:
    print(f"{m:7} | {N[m]:5} | {R[m]} | {Q[m]:.2f}")

total_reward = sum(sum(rlist) for rlist in R.values())
print(f"\nTotal Reward collected: {total_reward}")
```
